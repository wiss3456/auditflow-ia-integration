```python
def clean_raw_text(text):

    if not isinstance(text,str):
        return ''
    # remove Ui elements [content] [comments] and links (https:xxx)
    clean_text = re.sub(r'\[.*?\]|\(.*?\)','',text,flags=re.DOTALL)
    
    # removes stray ponctuation and artifacts
    clean_text = re.sub(r'(?<!\w)[!)\'"]+(?!\w)','',clean_text)

    # removing Non back space 
    clean_text =  clean_text.replace('\xa0',' ')

    #collapsing multiple spaces 
    clean_text = re.sub('[ \t]{2,}',' ',clean_text)

    # normalising multiple new lines 
    clean_text = re.sub(r'\n{3,}','\n\n',clean_text)
    
    clean_text = clean_text.replace('\\n', ' ')
    # cleanining the repeated title fragraments 
    clean_text = re.sub(r'["\']\s*\)', '', clean_text)

    # NEW: Remove TOC trailing dots and page numbers (e.g., "......22" or ". \n 42")
    clean_text = re.sub(r'\.{2,}\s*\d+', '', clean_text)

    # Removing ### and ** 
    clean_text = re.sub(r'(?<!\w)[#\*\|\-—\+]+(?!\w)','',clean_text)
    
    return clean_text.strip()


def clean_results(raw_results):
    # raw_results is a nested dict so first we access the second layers of the dict that have a list of dicts
    list_results = raw_results.get('results', [])
    # keeping the relevant results
    list_results = [r for r in list_results if r.get("score") > 0.5]
    for r in list_results:
        r['content'] = clean_raw_text(r.get('content'))
        r['raw_content'] = clean_raw_text(r.get('raw_content'))
    
    return list_results
```

### **simularity search using ``BM25`` model :** 

```python
from rank_bm25 import BM25Okapi

def rank_results(results_list,query):
    # stacking a list of title , content and raw content 
    texts = [
        ((r.get('title') or '') + r.get(('content') or '') + ' ' + (r.get("raw_content"))).split()
        for  r in results_list
    ]
    
    bm25 = BM25Okapi(texts)
    scores = bm25.get_scores(query.split())

    for i, r in enumerate(results_list):
        r['relevance_score'] = round(float(scores[i]),4)
    
    ranked = sorted(results_list,key= lambda x :x['relevance_score'],reverse=True)

    return ranked[:10]
```

### **Hybrid search system BM25 and embeddings**

```python
from sentence_transformers import SentenceTransformer 
import numpy as np 
model_emb = SentenceTransformer("all-MiniLM-L6-v2")
```

```python
def Ranking_results(search_results, search_query):
    Document = [
        (
            (r.get('title') or ' ') + 
        (r.get('content') or ' ') +
        (r.get('raw_content') or ' ')
        ) for r in search_results
    ]
    #--BM25
    tokenized = [doc.lower().split() for doc in Document]
    bm25 = BM25Okapi(tokenized)
    bm25_scores = bm25.get_scores(search_query.lower().split())
    bm25_ranks = np.argsort(bm25_scores)[::-1]

    #--embedding ranking
    Document_emb = model_emb.encode(Document,normalize_embeddings=True)
    query_emb = model_emb.encode(search_query,normalize_embeddings=True)
    emb_scores = np.dot(Document_emb,query_emb.T)
    emb_ranks = np.argsort(emb_scores)[::-1]

    #--RRF Reciprocal Ranking Fusion 
    k = 60 
    RRf_scores = {}
    # BM_scores :
    for rank , idx in enumerate(bm25_ranks):
        RRf_scores[idx] = RRf_scores.get(idx, 0) + 1/(k + rank)
    # Emb_scores :
    for rank , idx in enumerate(emb_ranks):
        RRf_scores[idx] = RRf_scores.get(idx, 0) + 1/(k + rank)

    final_rank = sorted(RRf_scores, key=RRf_scores.get, reverse=True)

    # Ranking the documents list 
    ranked_results = [search_results[int(i)] for i in final_rank]
    return ranked_results
```

##### **Document Chunking content :** 

```python
def chunk_document(
    document: dict,
    chunk_size=150,
    overlap=50
)-> list :
    """
    Split a document's raw_content into overlapping chunks.
    returns list of chunks 
    """
    source = document.get('raw_content')
    
    words = source.split()
    chunks = []
    start = 0
    while start < len(words):
        end = start + chunk_size
        chunk_words = words[start:end]
        chunk_text = " ".join(chunk_words)

        chunks.append(
            chunk_text,
        )
        start += chunk_size - overlap 

    return chunks 

doc = Ranked_results_2[0]
chunks = chunk_document(doc)
chunks
```

#### **Raw content Ranking via embedding :**

```python
def chunk_ranking(Documents:list ,top_k=5)-> list:
    """
    this functions Retrieve the relevant chunks from each document raw_content :
        - selecting top 5 Documents
        - Chunking each Document raw_content
        - Embedding each Document chunks 
        - sorting the chunks 
        - 
    """

    # selectig the Top 5 documents 
    top_Documents = Documents[:5]
    query_ref = reformulate_query("Query_reformation",query="Audit of Train Maintenance and Safety Management System")
    for doc in top_Documents:
        chunks = chunk_document(doc)
        chunks_emb = model_emb.encode(chunks,normalize_embeddings=True)
        query_emb = model_emb.encode(query_ref,normalize_embeddings=True)
        emb_score = np.dot(chunks_emb, query_emb.T)
        ranks = np.argsort(emb_score)[::-1]
        ranked_chunks = [chunks[int(i)] for i in ranks[:5]]
        doc['raw_content'] = ranked_chunks
        
    return top_Documents

final_documents = chunk_ranking(Ranked_results_2)
    
```
